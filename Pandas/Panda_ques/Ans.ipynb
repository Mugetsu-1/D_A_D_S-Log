```python
import pandas as pd
import numpy as np
```

Question 1: Create DataFrame from Dictionary


```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'Age': [25, 30, 35, 28, 32],
    'Department': ['HR', 'IT', 'Finance', 'IT', 'HR'],
    'Salary': [50000, 65000, 70000, 62000, 55000]
}

df = pd.DataFrame(data)
print(df)
```

Question 2: Read CSV File


```python
# Creating a sample CSV first
df.to_csv('sample_data.csv', index=False)

# Read and display first 5 rows
csv_df = pd.read_csv('sample_data.csv')
print(csv_df.head())
```

Question 3: Get DataFrame Shape


```python
print(f"Shape: {df.shape}")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

Question 4: Display Columns and Data Types


```python
print("Column names:", df.columns.tolist())
print("\nData types:")
print(df.dtypes)
```

Question 5: Column Selection


```python
# Single column
name_col = df['Name']
print("Single column 'Name':")
print(name_col)

# Multiple columns
name_age = df[['Name', 'Age']]
print("\nMultiple columns 'Name' and 'Age':")
print(name_age)
```

Question 6: Row Selection with iloc


```python
first_three = df.iloc[:3]
print("First 3 rows:")
print(first_three)
```

Question 7: Find Missing Values


```python
# Create DataFrame with missing values
df_with_nulls = df.copy()
df_with_nulls.loc[2, 'Age'] = np.nan
df_with_nulls.loc[0, 'Department'] = np.nan

print("DataFrame with missing values:")
print(df_with_nulls)

print("\nMissing values count:")
print(df_with_nulls.isna().sum())

print("\nRows with any missing values:")
print(df_with_nulls[df_with_nulls.isna().any(axis=1)])
```

Question 8: Replace NaN Values


```python
# Replace NaN in 'Age' with mean, 'Department' with 'Unknown'
df_filled = df_with_nulls.copy()
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].mean())
df_filled['Department'] = df_filled['Department'].fillna('Unknown')

print("After replacing NaN values:")
print(df_filled)
```

Question 9: Sort DataFrame


```python
# Sort by Age descending
df_sorted = df.sort_values('Age', ascending=False)
print("Sorted by Age (descending):")
print(df_sorted)

# Multi-column sort
df_multi_sort = df.sort_values(['Department', 'Age'], ascending=[True, False])
print("\nSorted by Department (asc) and Age (desc):")
print(df_multi_sort)
```

Question 10: GroupBy Operations


```python
# Group by Department and calculate average salary
grouped = df.groupby('Department')['Salary'].mean()
print("Average salary by department:")
print(grouped)

# Multiple aggregations
agg_results = df.groupby('Department').agg({
    'Salary': ['mean', 'min', 'max', 'count'],
    'Age': 'mean'
})
print("\nMultiple aggregations:")
print(agg_results)
```

Imputation Techniques

Imputation is the process of replacing missing data with substituted values.

1) Mean Value Imputation

Handling missing values by replacing Missing Values with Mean

```python
sns.histplot(df['age'], kde=True) # display a histogram of the 'age' column
```

```python
df['age']
```

```python
df['age'].isnull().sum() # display the number of missing values in the 'age' column
```

```python
df['age'].mean() # display the mean of the 'age' column
```

```python
df
```

```python
df['age'].mean()
```

```python
df['Age_mean'] = df['age'].fillna(df['age'].mean()) # fill missing values with the mean of the 'age' column
```

```python
df[['Age_mean', 'age']] # display the 'Age_mean' and 'age' columns
```

```python
df['Age_mean'].isna().sum()
```

This technique works well when the data is normally distributed.

2) Median Value Imputation

Handling missing values by replacing Missing Values with Median

If you have outliers in dataset use this technique.

```python
df['age'].median() # display the median of the 'age' column
```

```python
df.sample(3)
```

```python
df['Age_median'] = df['age'].fillna(df['age'].median()) # fill missing values with the median of the 'age' column
```

```python
df[['Age_median', 'Age_mean', 'age']] # display the 'Age_median', 'Age_mean', and 'age' columns
```

```python
df['Age_median'].isna().sum() # display the number of missing values in the 'Age_median' column
```

```python
df['Age_mean'].isna().sum() # display the number of missing values in the 'Age_mean' column
```

```python
df['age'].isna().sum() # display the number of missing values in the 'age' column
```

```python
type(df[df['embarked'].notna()]['embarked'].mode())
```

```python
df[df['embarked'].notna()]['embarked'].mode()[0] # display the first mode of the 'embarked' column for rows with non-missing values in the 'embarked' column
```

```python
mode = df[df['embarked'].notna()]['embarked'].mode()[0] # store the mode of the 'embarked' column for rows with non-missing values in the 'embarked' column
```

```python
mode
```

```python
df['embarked_mode'] = df['embarked'].fillna(mode) # fill missing values in the 'embarked' column with the mode
```

```python
df['embarked_mode'].isnull().sum() # display the number of missing values in the 'embarked_mode' column
```

```python
df['embarked'].isnull().sum() # display the number of missing values in the 'embarked' column
```

1. Basic Plotting Functions


Line Plot

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, color='r', linestyle='--', linewidth=2, 
         marker='o', markersize=5, label='sin(x)')
plt.legend()
plt.show()
```

Scatter Plot


```python
plt.scatter(x, y, c='blue', s=50, marker='s', 
           alpha=0.7, edgecolor='black')
plt.show()
```

Bar Chart


```python
categories = ['A', 'B', 'C', 'D']
values = [15, 23, 12, 18]

plt.bar(categories, values, width=0.6, 
        color='skyblue', edgecolor='black')
plt.show()
```

Histogram


```python
data = np.random.randn(1000)
plt.hist(data, bins=20, color='green', 
         alpha=0.7, edgecolor='black')
plt.show()
```

Pie Chart


```python
labels = ['Python', 'Java', 'C++', 'JavaScript']
sizes = [40, 25, 20, 15]
explode = (0.1, 0, 0, 0)  # explode 1st slice

plt.pie(sizes, labels=labels, explode=explode,
        autopct='%1.1f%%', shadow=True)
plt.show()
```

2. Figure & Axis Management


```python
# Create figure with specific size
plt.figure(figsize=(10, 6))

# Multiple subplots method 1
plt.subplot(2, 2, 1)  # 2 rows, 2 cols, position 1
plt.plot(x, y)

plt.subplot(2, 2, 2)
plt.scatter(x, y)

# Method 2: Using subplots
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes[0, 0].plot(x, y)
axes[0, 1].scatter(x, y)
axes[1, 0].bar(categories, values)
axes[1, 1].hist(data, bins=20)

# Set axis limits
plt.axis([0, 10, -1.5, 1.5])
# Or separately
plt.xlim(0, 10)
plt.ylim(-1.5, 1.5)

# Customize ticks
plt.xticks([0, 5, 10], ['Start', 'Middle', 'End'])
plt.yticks([-1, 0, 1], ['Low', 'Zero', 'High'])

# Tick parameters
plt.tick_params(axis='both', direction='in', length=6, 
                width=2, colors='red', labelsize=10)
```

3. Labels, Titles & Legends


```python
plt.plot(x, y, label='Data Series')

# Labels and title
plt.xlabel('X-axis Label', fontsize=12, color='blue')
plt.ylabel('Y-axis Label', fontsize=12, color='blue')
plt.title('My Plot Title', fontsize=14, color='darkred')

# Legend with custom location
plt.legend(loc='upper right', fontsize=10, 
           frameon=True, shadow=True)

# Grid
plt.grid(True, linestyle='--', alpha=0.5, color='gray')
```

4. Complete Example


```python
import matplotlib.pyplot as plt
import numpy as np

# Create data
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

# Create figure with subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

# First subplot: Line plot
ax1.plot(x, y1, 'b-', label='sin(x)', linewidth=2, alpha=0.8)
ax1.plot(x, y2, 'r--', label='cos(x)', linewidth=2, alpha=0.8)
ax1.set_xlabel('X values')
ax1.set_ylabel('Y values')
ax1.set_title('Trigonometric Functions')
ax1.legend(loc='best')
ax1.grid(True, alpha=0.3)

# Second subplot: Scatter plot
ax2.scatter(x, y1, c='blue', s=20, label='sin(x)', alpha=0.6)
ax2.scatter(x, y2, c='red', s=20, label='cos(x)', alpha=0.6)
ax2.set_xlabel('X values')
ax2.set_ylabel('Y values')
ax2.set_title('Scatter Plot')
ax2.legend()
ax2.grid(True, alpha=0.3)

# Adjust layout
plt.tight_layout()

# Save figure
plt.savefig('my_plot.png', dpi=300, bbox_inches='tight')

# Show plot
plt.show()
```

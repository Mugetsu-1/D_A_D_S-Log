1. How do you plot a simple line graph?


```python
import matplotlib.pyplot as plt
plt.plot([1, 2, 3, 4], [1, 4, 9, 16])
plt.show()
```

2. Difference between plt.plot() and plt.scatter()

>  plt.plot(): Connects points with lines, faster for large datasets

>  plt.scatter(): Shows individual points, can vary color/size per point

3. Change line color and style


```python
plt.plot(x, y, color='red', linestyle='--', linewidth=2)
# or shorthand: plt.plot(x, y, 'r--', linewidth=2)
```

4. Set axis labels


```python
plt.xlabel('X-axis Label')
plt.ylabel('Y-axis Label')
```

7. Alpha parameter

8. Multiple subplots


```python
fig, axes = plt.subplots(2, 2)  # 2x2 grid
axes[0, 0].plot(x, y)  # Top-left
axes[0, 1].scatter(x, y)  # Top-right
```

9. Difference between plt.bar() and plt.barh()

>  plt.bar(): Vertical bars (x-position, height)

>  plt.barh(): Horizontal bars (y-position, width)

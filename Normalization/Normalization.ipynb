Transformation --> xscaled = (x - xmin) / (xmax - xmin)

Mostly use standardization in Machine learning and normalization in deep learning.

```python
import seaborn as sns
```

```python
df = sns.load_dataset('taxis')
df.head()
```

```python
from sklearn.preprocessing import MinMaxScaler # importing MinMaxScaler from sklearn.preprocessing 
```

```python
min_max = MinMaxScaler()
```

```python
min_max
```

```python
df[['distance', 'fare', 'tip']]
```

```python
max(df['distance']), min(df['distance']), max(df['fare']), min(df['fare']), max(df['tip']), min(df['tip']) # checking the max and min values of the columns
```

```python
min_max.fit(df[['distance', 'fare', 'tip']]) # fitting the data to the MinMaxScaler
```

```python
min_max.transform(df[['distance', 'fare', 'tip']]) # transforming the data to the MinMaxScaler
```

```python
min_max.fit_transform(df[['distance', 'fare', 'tip']]) # fitting and transforming the data to the MinMaxScaler
```

```python
import pandas as pd
df = pd.DataFrame(min_max.fit_transform(df[['distance', 'fare', 'tip']]), columns=['distance', 'fare', 'tip']) # creating a new dataframe with the transformed data
```

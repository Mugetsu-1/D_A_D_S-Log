```python
import requests
from bs4 import BeautifulSoup
```

```python
web = requests.get("https://webscraper.io/test-sites/e-commerce/allinone/computers/laptops")
```

```python
soup = BeautifulSoup(web.content, 'html.parser')

names = soup.find_all("a", class_ = "title")
na = []
for name in names:
  print(name.text)
  na.append(name.text.strip())
```

```python
prices = soup.find_all("h4", class_="price")
pr = []
for price in prices:
  print(price.text)
  pr.append(price.text.strip())
```

```python
descriptions = soup.find_all("p", class_="description")
desc = []
for description in descriptions:
  print(description.text)
  desc.append(description.text.strip())
```

```python

ratings = soup.find_all("p",{'data-rating': True})
rate = []
for i in ratings:
  r = i.get("data-rating")
  rate.append(r)
  print(r)
```

```python
reviews = soup.find_all("span",{"itemprop":"reviewCount"})
review = []
for i in reviews:
  review.append(i.text.strip())
  print(i.text)
```

```python
data = {
    "Name":na,
    "Description":desc,
    "Rating":rate,
    "Reviews":review
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(data)
df.head()
```

```python
df.to_csv("CSVs/Ecommerce.csv",index=False)
```

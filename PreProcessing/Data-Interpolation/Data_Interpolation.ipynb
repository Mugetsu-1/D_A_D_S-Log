**Data Interpolation**

Data Interpolation is the process of estimating unknown values within a dataset based on the known values. In Python, there are various libraries available that can be used for data interpolation, such as NumPy, SciPy, and Pandas. Here is an example of how to perform data interpolation using the NumPy library :

1. Linear Interpolation

The simplest and most commonly used.

Connects two known points with a straight line and estimates intermediate values.

| Type               | Description                  | Smoothness | Use Case                  |
| ------------------ | ---------------------------- | ---------- | ------------------------- |
| Linear             | Straight-line between points | Moderate   | Time-series gaps          |
| Polynomial         | Polynomial curve fit         | High       | Continuous smooth data    |
| Spline             | Piecewise smooth polynomial  | Very High  | Smooth curves             |
| Nearest            | Closest value used           | Low        | Image or categorical data |
| Piecewise Constant | Step-like pattern            | Low        | Discrete data             |
| Lagrange           | Polynomial using basis       | High       | Small datasets            |


```python
import numpy as np
x = np.array([1,2,3,4,5])
y = np.array([2,4,6,8,10])
```

```python
import matplotlib.pyplot as plt
plt.scatter(x,y)
```

```python
## Interpolating the data using linear interpolation

# Create 10 evenly spaced new x-values between 1 and 5
# np.linspace(start, stop, num) → generates 'num' equally spaced points from 'start' to 'stop'
x_new = np.linspace(1, 5, 10)

# Perform linear interpolation to estimate y-values for the new x-values
# np.interp(new_x, old_x, old_y)
# - new_x: x-values where we want to estimate y
# - old_x: original x-values
# - old_y: original y-values
# The function connects existing points with straight lines and estimates intermediate values
y_interp = np.interp(x_new, x, y)

# Print the interpolated y-values (estimated points)
print(y_interp)

# Print the new x-values created for interpolation
print(x_new)
```

```python
plt.scatter(x_new,y_interp)
```

 Cubic Interpolation with Scipy =>  smooth, curvy line between points

 CUbic Interpolation is a type of Polynomial Interpolation

```python
x = np.array([1,2,3,4,5])
y = np.array([1,8,27,64,125])
```

```python
from scipy.interpolate import interp1d
```

```python
# Create a cubic interpolation function
f = interp1d(x,y,kind = 'linear')
```

```python
# Interpolate the data
x_new = np.linspace(1,5,10)
y_interp = f(x_new)
print(y_interp)
```

```python
plt.scatter(x,y)
```

```python
plt.scatter(x_new, y_interp)
```

```python
# Import NumPy
import numpy as np
```

Question 1: Create a NumPy array of numbers from 10 to 50


```python
array_10_to_50 = np.arange(10, 51)
print("Array from 10 to 50:")
print(array_10_to_50)
print(f"Shape: {array_10_to_50.shape}")
print(f"Type: {array_10_to_50.dtype}")
```

Question 2: Generate a 3×3 identity matrix


```python
identity_matrix = np.eye(3)
print("3x3 Identity Matrix:")
print(identity_matrix)
print(f"Shape: {identity_matrix.shape}")
```

Question 3: Create a NumPy array of 10 random integers between 1 and 100


```python
np.random.seed(42)  # For reproducibility
random_integers = np.random.randint(1, 101, 10)
print("10 random integers between 1 and 100:")
print(random_integers)
print(f"Length: {len(random_integers)}")
```

Question 4: Find the maximum and minimum of an array


```python
sample_array = np.array([23, 45, 12, 67, 89, 34, 56, 78, 91, 14])
max_value = np.max(sample_array)
min_value = np.min(sample_array)
max_value_alt = sample_array.max()
min_value_alt = sample_array.min()

print("Sample array:", sample_array)
print(f"Maximum value: {max_value} (using np.max)")
print(f"Minimum value: {min_value} (using np.min)")
print(f"Maximum value: {max_value_alt} (using .max() method)")
print(f"Minimum value: {min_value_alt} (using .min() method)")
```

Question 5: Reshape a 1D array of 12 elements into a 3×4 matrix


```python
# First create a 1D array of 12 elements
array_1d = np.arange(1, 13)
print("Original 1D array:")
print(array_1d)
print(f"Shape: {array_1d.shape}")

# Reshape to 3x4
array_3x4 = array_1d.reshape(3, 4)
print("\nReshaped to 3x4 matrix:")
print(array_3x4)
print(f"Shape: {array_3x4.shape}")

# Alternative method
array_3x4_alt = np.reshape(array_1d, (3, 4))
print("\nAlternative method - same result:")
print(array_3x4_alt)
```

Question 6: Create an array of all even numbers between 50 and 100


```python
even_numbers = np.arange(50, 101, 2)
print("Even numbers between 50 and 100:")
print(even_numbers)
print(f"Count: {len(even_numbers)} even numbers")

#Alternative: Using boolean indexing
all_numbers = np.arange(50, 101)
even_numbers_alt = all_numbers[all_numbers % 2 == 0]
print("\nAlternative method (using boolean indexing):")
print(even_numbers_alt)
```

Question 7: Given an array [1,2,3,4,5], compute the square of each element


```python
original_array = np.array([1, 2, 3, 4, 5])
squared_array = original_array ** 2
print(f"Original array: {original_array}")
print(f"Squared array: {squared_array}")

#Alternative methods
squared_array_np = np.square(original_array)
squared_array_mult = original_array * original_array

print(f"\nUsing np.square(): {squared_array_np}")
print(f"Using element-wise multiplication: {squared_array_mult}")
```

Question 8: Create a 5×5 array of random numbers and find its mean


```python
np.random.seed(42)  # For reproducibility
random_5x5 = np.random.rand(5, 5)
mean_value = np.mean(random_5x5)

print("5x5 random array (values between 0 and 1):")
print(random_5x5)
print(f"\nMean of the array: {mean_value:.4f}")

# Additional statistics
print(f"Row-wise means: {np.mean(random_5x5, axis=1).round(4)}")
print(f"Column-wise means: {np.mean(random_5x5, axis=0).round(4)}")
```

Question 9: Find the index of the maximum element in [10, 20, 50, 40, 30]


```python
array_to_search = np.array([10, 20, 50, 40, 30])
max_index = np.argmax(array_to_search)

print(f"Array: {array_to_search}")
print(f"Maximum value: {array_to_search[max_index]}")
print(f"Index of maximum value: {max_index}")

# What if there are multiple maximum values?
array_with_duplicates = np.array([10, 50, 30, 50, 20])
max_indices = np.where(array_with_duplicates == np.max(array_with_duplicates))[0]
print(f"\nArray with duplicates: {array_with_duplicates}")
print(f"Indices of maximum value (50): {max_indices}")
```

Question 10: Create two arrays and perform element-wise operations


```python
#Create two arrays
array_a = np.array([10, 20, 30, 40, 50])
array_b = np.array([1, 2, 3, 4, 5])

print(f"Array A: {array_a}")
print(f"Array B: {array_b}")

# Element-wise operations
addition = array_a + array_b
subtraction = array_a - array_b
multiplication = array_a * array_b
division = array_a / array_b

print("\nElement-wise operations:")
print(f"Addition (A + B): {addition}")
print(f"Subtraction (A - B): {subtraction}")
print(f"Multiplication (A * B): {multiplication}")
print(f"Division (A / B): {division}")

# Using NumPy functions
print("\nUsing NumPy functions:")
print(f"np.add(): {np.add(array_a, array_b)}")
print(f"np.subtract(): {np.subtract(array_a, array_b)}")
print(f"np.multiply(): {np.multiply(array_a, array_b)}")
print(f"np.divide(): {np.divide(array_a, array_b)}")
```

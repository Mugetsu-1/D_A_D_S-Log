Inferential statistics is the branch of statistics that allows us to make conclusions, predictions, or generalizations about a population based on data collected from a sample.

In simple terms — instead of studying every individual in a population (which is often impossible), we analyze a smaller sample and infer (estimate) what the population might look like.

| **Basis**           | **Descriptive Statistics**                                   | **Inferential Statistics**                                                   |
| ------------------- | ------------------------------------------------------------ | ---------------------------------------------------------------------------- |
| **Definition**      | Describes and summarizes data from a sample or population.   | Makes predictions or generalizations about a population based on a sample.   |
| **Purpose**         | To present data in a meaningful and understandable way.      | To draw conclusions and make decisions from data.                            |
| **Data Type**       | Uses actual, observed data.                                  | Uses sample data to infer about population.                                  |
| **Techniques Used** | Mean, median, mode, standard deviation, charts, graphs, etc. | Hypothesis testing, confidence intervals, regression, ANOVA, etc.            |
| **Nature**          | Deals with what has happened.                                | Deals with what may happen or could be true.                                 |
| **Example**         | Average marks of students in a class.                        | Estimating the average marks of all students in a country based on a sample. |


#Key Concepts

##Population vs Sample

Population: The entire group you want to study (e.g., all students in Nepal).

Sample: A smaller group taken from the population (e.g., 500 randomly selected students).

##Sampling

The process of selecting a sample from the population.

Proper sampling ensures the results are representative and unbiased.

##Estimation

Using sample data to estimate population parameters (like mean, variance, or proportion).

Example: Estimating the average height of all Nepali adults based on a sample.

##Hypothesis Testing

A formal method to test assumptions about population parameters.

Example: “Does a new teaching method improve students’ performance?”

##Confidence Intervals

A range of values used to estimate a population parameter with a certain level of confidence (e.g., 95%).

##Regression & Correlation

Techniques to find relationships between variables and predict future outcomes.

Estimate

Point Estimate: Single numerical value used to estimate the unknown population parameter.

Sample mean is the point estimate of the population mean.

Interval Estimate: Range of values that is used to estimate the unknown population parameter.

Interval estimate of population parameters are called cofidence intervals.

it’s not just about probability in general, but specifically about how probabilities change when we get new evidence.

##Probability

Probability is the chance that something will happen.

Example:

Tossing a coin → chance of getting heads = 1 out of 2.

Rolling a die → chance of getting 6 = 1 out of 6.

In short: Probability = Likelihood of an event.

##Dependent Events

Two events are dependent if the outcome of one affects the outcome of the other.

In other words, knowing that one event happened changes the probability of the second event.

Example: Tossing a coin twice.

First toss: Heads

Second toss: ?

Outcome of the first toss does not affect the second toss.

In short:

“The first event changes the chance of the second.”

##Independent Events

Two events are independent if the outcome of one does not affect the outcome of the other.

The occurrence of one has no impact on the probability of the other.

Example: Grabbing balls from a bag without replacement.

Bag has 3 red and 2 blue balls.

First ball drawn: Red

Second ball: Probability of drawing red changes because one red is already taken.

In short:

“One event has no effect on the other.”

##Conditional Probability
It is somehow similar to dependent events.

Bag with balls: picking a red ball first affects the chance of picking another red → dependent events.

Calculating the probability of the second red ball → conditional probability.

##Example:

A bag has 3 red and 2 blue balls.

You pick one ball, and it is red.

Now, the probability of picking another red ball depends on the first pick.

This is conditional probability because the second event depends on the first.

💡 Simple way to remember:

It’s the chance of something happening given that something else has already happened.

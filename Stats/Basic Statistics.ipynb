**Types of Statistics**

- **Descriptive Statistics**
- It consists of organizing and summarizing data.
- Measure of Central Tendency (Mean, Median and Mode)
- Measure of Dispersion (Variance and Standard Deviation)
- Variance and Standard Deviation tell us how spread out the data is from the mean.
- Different types of distribution of data (Histogram, pdf( Probabolity Densoty Function), pmf(Probability Mass Function))

Quick analogy

Histogram: Picture of your data’s shape.

PMF: Probability table for countable outcomes.

PDF: Smooth curve describing probability for measurable outcomes.



1. **Histogram** → raw counts from data.
2. **PMF** → normalized histogram for discrete values (probabilities).
3. **PDF** → smooth probability curve for continuous values, estimated from data.

It’s essentially one concept evolving across **raw → discrete → continuous** cases.


- **Inferential Statistics**

- It consists of using data that we have collected to form conclusions. (Use sample data to draw conclusion for population.)
- Z test, t test, Chi square test (Hypothesis Testing)

Descriptive Statistics → Just describe what your data looks like right now. (Example: "The average height of students in this class is 165 cm.")

Inferential Statistics → Use a small sample of data to make predictions or conclusions about a larger group or future. (Example: "Based on this sample of 50 students, we predict the average height of all students in the school is about 165 cm.")

For ex : Let's say there are 20 classes in the college and you have collected the heights of students in a class.

Heights are recorded as 175 cm, 180 cm, 140 cm, 140 cm, 135 cm, 160 cm, 135 cm, 190 cm

Descriptive Question : "What is the average height of the entire classroom?"

Inferential Question : "Are the height of students in classroom are similar to what you expect in the entire college?"

**Measure of Dispersion** --> Spreadness of the data

- Variance
- Standard Deviation

Population Variance

x - Data Points
μ - Population Mean
N - Population Size

Sample Variance

x - Data Points x̄ - Sample Mean n - Sample Size

Why we divide sample variance by n - 1?

It is done so that we can create an unbiased estimator of the population variance.

For ex : {1, 2, 3, 4,5}

|  x  | x̄     | (x - x̄)     | (x - x̄)^2     |
| --- | ----- | ----------- | ------------- |
|  1  |   3   |     -2      |      4        |
|  2  |   3   |     -1      |      1        |
|  3  |   3   |      0      |      0        |
|  4  |   3   |      1      |      1        |
|  5  |   3   |      2      |      4        |


Sample Variance = 2.5

X = {.......}                  y = {........}

Sample Variance of X = 2.5(Less Spread) 

Sample Variance of y = 7.5(More Spread)

So, variance tells about dispersion or spreadness of data.

Standard Deviation

- Population Standard Deviation (σ = √variance)
- Sample Standard Deviation (√sample variance)

- X = {1, 2, 3, 4, 5}
- x̄ = 3
- σ = 1

One step towards left and one step towards right in normal distribution

**Random Variable**

x + 5 = 7                            

8 = y + x

x and y are variable (x=2 and y = 6)

Random variable is a process of mapping the output of a random process or experiments to a number.

For ex : Tossing a coin

X = {0 if Head and 1 if Tails}

Rolling a dice

y = {sum of rolling of dice 7 times}

```python
# Coin Toss Example
import numpy as np

# Simulate 10 coin tosses (0 = Head, 1 = Tail)
coin_tosses = np.random.choice([0, 1], size=10)

print("Coin toss outcomes:", coin_tosses)
print("P(X = 1) → Probability of Tail:", np.mean(coin_tosses))
print("P(X = 0) → Probability of Head:", 1 - np.mean(coin_tosses))
```

| `np.in1d(arr1, arr2)`              | Boolean array: `True` if element of `arr1` is in `arr2`.          | `np.in1d([1, 2, 3], [2, 4])`            | `[False  True False]` |
| `np.isin(elements, test_elements)` | Tests if elements exist in test array (works for scalars/arrays). | `np.isin([1, 2, 3], [2, 4])`            | `[False  True False]` |

**Histogram and Skewness**

Histogram gives idea about frequency.
A histogram is a bar graph where each bar represents a range (or "bin") of values, and the height of the bar indicates the frequency (or count) of data points within that bin.
 Why Histograms Matter in ML:
Feature Distribution Analysis:

Check if a feature is normally distributed, skewed, or uniform.

Helps decide whether to apply scaling, log transformation, or normalization.

Outlier Detection:

Rare values with very low frequencies may show up in the tails of the histogram.

Handling Skewness:

Right or left-skewed features may affect models like linear regression or SVM. Histograms help spot them.

Bin Selection:

Proper bin size ensures a balance between over-smoothing and overfitting to noise in visualization.

Normal/Gaussian Distribution

Key Things It Tells Us:

1️⃣ Shape of Data

It’s bell-shaped and symmetrical.

Most of the data is concentrated near the mean (center), with fewer values farther away.

2️⃣ Mean = Median = Mode

In a perfect normal distribution, the mean, median, and mode are all the same.

3️⃣ Probability & the 68–95–99.7 Rule

It tells us how much of the data lies within certain standard deviations (
𝜎
σ) from the mean (
𝜇
μ):

The standard normal distribution is just a special case of the normal (Gaussian) distribution.

- 1st sd = 68.2 %
- 2nd sd = 95.4 %
- 3rd sd = 99.6 %

Symmetrical Distribution ---> No Skewness

The mean, median and mode all are perfectly at the center.

Mean = Median = Mode

Q3 - Q2 ≈ Q2 - Q1

Right Skewness (Positive Skew)

Tail is stretched to the right (towards larger values).

The mean > median > mode (because the mean gets pulled in the direction of the long tail).

Left Skewness (Negative Skew)

Tail is stretched to the left (towards smaller values).

The mean < median < mode (because the mean gets pulled left by small outliers).

Relationship between X and Y

What is the relationship between X and Y?

Covariance and correlation are used to quantify the relation.

There are four conditions 
- When X increases, Y also increases.
- When X increases, Y decreases.
- When X decreases, Y also decreases.
- When X decreases, Y increases.

When X is increasing Y is also increasing. When X is decreasing Y is also decreasing.

Relationship between size of house and price of house?

- xi --> Datapoints of x
- x̄ --> Sample mean
- yi --> Datapoints of y
- ȳ --> Sample mean of y

Var(X) = Cov(X,X)

When X increases, Y increases and when X decreases, Y decreases. This is positive covariance.

When X increases, Y decreases and when X decreases, Y increases. This is negative covariance.

Positive Value --> Positive Covariance

Advantage:
- Relationship between X and Y.

Disadvantage:
- Covariance does not have a specific limit value.

**Pearson Correlation Coefficient** [-1 to +1]

- The more the value towards +1 the more +ve correlated it is.
- The more the value towards -1 the more -ve correlated it is.
